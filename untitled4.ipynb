```python
import pandapower as pp
import pandapower.shortcircuit as sc
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import numpy as np

# ==============================================================================
# 1. ŞEBEKE MODELİNİN OLUŞTURULMASI (14 Baralı OG + AG PV & Rüzgar Sistemi)
# ==============================================================================
net = pp.create_empty_network(name="14_Barali_PV_Ruzgar_OG_AG_Sebekesi", f_hz=50.0)

# Baralar (154 kV İletim, 33 kV OG, 0.4 kV AG)
b0  = pp.create_bus(net, vn_kv=154.0, name="B0_154kV_Iletim")
b1  = pp.create_bus(net, vn_kv=33.0,  name="B1_33kV_AnaBara")

# OG Fider 1
b2  = pp.create_bus(net, vn_kv=33.0, name="B2_OG_Fider1")
b3  = pp.create_bus(net, vn_kv=33.0, name="B3_OG_Fider1")
b4  = pp.create_bus(net, vn_kv=33.0, name="B4_OG_Fider1")
b5  = pp.create_bus(net, vn_kv=33.0, name="B5_OG_Ruzgar_Bara")

# OG Fider 2
b6  = pp.create_bus(net, vn_kv=33.0, name="B6_OG_Fider2")
b7  = pp.create_bus(net, vn_kv=33.0, name="B7_OG_Fider2")
b8  = pp.create_bus(net, vn_kv=33.0, name="B8_OG_Fider2")
b9  = pp.create_bus(net, vn_kv=33.0, name="B9_OG_PV_Bara")
b10 = pp.create_bus(net, vn_kv=33.0, name="B10_OG_SonBara")

# AG Baraları (0.4 kV)
b11 = pp.create_bus(net, vn_kv=0.4, name="B11_AG_Konut1")
b12 = pp.create_bus(net, vn_kv=0.4, name="B12_AG_CatiPV")
b13 = pp.create_bus(net, vn_kv=0.4, name="B13_AG_Konut2")

# Harici Şebeke (154 kV Slack)
pp.create_ext_grid(net, bus=b0, vm_pu=1.02, va_degree=0.0,
                   s_sc_max_mva=2500.0, s_sc_min_mva=1800.0,
                   rx_max=0.1, rx_min=0.1, name="154kV_Ulusal_Sebeke")

# Ana İndirici Trafo (154/33 kV - 40 MVA)
pp.create_transformer_from_parameters(
    net, hv_bus=b0, lv_bus=b1, sn_mva=40.0, vn_hv_kv=154.0, vn_lv_kv=33.0,
    vkr_percent=0.45, vk_percent=12.0, pfe_kw=25.0, i0_percent=0.15,
    vector_group="YNyn0", vk0_percent=12.0, vkr0_percent=0.45, mag0_percent=100, mag0_rx=0.0,
    name="TR1_Ana_154_33kV_40MVA"
)

# Dağıtım Trafoları (33/0.4 kV - 630 kVA)
for hv, lv, t_name in [
    (b4, b11, "TR2_Dagitim_33_04kV_630kVA"),
    (b9, b12, "TR3_Dagitim_33_04kV_630kVA"),
    (b10, b13, "TR4_Dagitim_33_04kV_630kVA")
]:
    pp.create_transformer_from_parameters(
        net, hv_bus=hv, lv_bus=lv, sn_mva=0.63, vn_hv_kv=33.0, vn_lv_kv=0.4,
        vkr_percent=1.2, vk_percent=6.0, pfe_kw=1.1, i0_percent=0.25,
        vector_group="Dyn11", vk0_percent=6.0, vkr0_percent=1.2, mag0_percent=100, mag0_rx=0.0,
        name=t_name
    )

# Hat Tipleri (33 kV Havai Hat ve Yeraltı Kablosu)
pp.create_std_type(net, {
    "r_ohm_per_km": 0.157, "x_ohm_per_km": 0.350, "c_nf_per_km": 9.5,
    "max_i_ka": 0.535, "type": "ol"
}, name="33kV_Havai_Hat_3_0_AWG", element="line")

pp.create_std_type(net, {
    "r_ohm_per_km": 0.164, "x_ohm_per_km": 0.118, "c_nf_per_km": 210.0,
    "max_i_ka": 0.365, "type": "cs"
}, name="33kV_Yeralti_Kablosu_1x185", element="line")

# Fider 1 Hatları
pp.create_line(net, b1, b2, length_km=3.5, std_type="33kV_Yeralti_Kablosu_1x185", name="H1_B1_B2_Kablo")
pp.create_line(net, b2, b3, length_km=4.2, std_type="33kV_Havai_Hat_3_0_AWG", name="H2_B2_B3_Havai")
pp.create_line(net, b3, b4, length_km=3.8, std_type="33kV_Havai_Hat_3_0_AWG", name="H3_B3_B4_Havai")
pp.create_line(net, b4, b5, length_km=5.0, std_type="33kV_Havai_Hat_3_0_AWG", name="H4_B4_B5_Havai")

# Fider 2 Hatları
pp.create_line(net, b1, b6, length_km=2.8, std_type="33kV_Yeralti_Kablosu_1x185", name="H5_B1_B6_Kablo")
pp.create_line(net, b6, b7, length_km=4.5, std_type="33kV_Havai_Hat_3_0_AWG", name="H6_B6_B7_Havai")
pp.create_line(net, b7, b8, length_km=3.2, std_type="33kV_Yeralti_Kablosu_1x185", name="H7_B7_B8_Kablo")
pp.create_line(net, b8, b9, length_km=4.0, std_type="33kV_Havai_Hat_3_0_AWG", name="H8_B8_B9_Havai")
pp.create_line(net, b9, b10, length_km=3.6, std_type="33kV_Havai_Hat_3_0_AWG", name="H9_B9_B10_Havai")

net.line["endtemp_degree"] = 80.0

# Yükler (cos_phi ≈ 0.95 kabulü ile)
pp.create_load(net, bus=b2, p_mw=2.2, q_mvar=0.70, name="Yuk_B2_OG_Sanayi1")
pp.create_load(net, bus=b3, p_mw=1.8, q_mvar=0.55, name="Yuk_B3_OG_Ticari")
pp.create_load(net, bus=b5, p_mw=1.5, q_mvar=0.45, name="Yuk_B5_OG_Sanayi2")
pp.create_load(net, bus=b6, p_mw=2.0, q_mvar=0.65, name="Yuk_B6_OG_Sanayi3")
pp.create_load(net, bus=b7, p_mw=1.6, q_mvar=0.50, name="Yuk_B7_OG_Tarimsal")
pp.create_load(net, bus=b8, p_mw=2.4, q_mvar=0.80, name="Yuk_B8_OG_Sanayi4")
pp.create_load(net, bus=b10, p_mw=1.4, q_mvar=0.45, name="Yuk_B10_OG_Ticari2")

pp.create_load(net, bus=b11, p_mw=0.45, q_mvar=0.14, name="Yuk_B11_AG_Konut1")
pp.create_load(net, bus=b12, p_mw=0.40, q_mvar=0.12, name="Yuk_B12_AG_Konut2")
pp.create_load(net, bus=b13, p_mw=0.48, q_mvar=0.15, name="Yuk_B13_AG_Konut3")

# Yenilenebilir Enerji Kaynakları (PV + Rüzgar)
pp.create_sgen(net, bus=b5, p_mw=4.0, q_mvar=0.0, sn_mva=4.5, k=1.2, type="WP", name="Ruzgar_Turbini_4MW")
pp.create_sgen(net, bus=b9, p_mw=3.0, q_mvar=0.0, sn_mva=3.3, k=1.2, type="PV", name="PV_Santrali_OG_3MW")
pp.create_sgen(net, bus=b12, p_mw=0.25, q_mvar=0.0, sn_mva=0.28, k=1.2, type="PV", name="Cati_PV_AG_250kW")

# 1. Proje Teslim Dosyasını Kaydet
pp.to_json(net, "14_barali_pv_ruzgar_sebekesi.json")

# ==============================================================================
# 2. TEK HAT ŞEMASI ÇİZİMİ (14 BARANIN TAMAMI VE TÜM ETİKETLERLE)
# ==============================================================================
coords = {
    0: (0.5, 10), 1: (2.5, 10),
    2: (4.5, 13), 3: (6.5, 13), 4: (8.5, 13), 5: (10.5, 13),
    6: (4.5, 7),  7: (6.5, 7),  8: (8.5, 7),  9: (10.5, 7), 10: (12.5, 7),
    11: (8.5, 16), 12: (10.5, 4), 13: (12.5, 4)
}

fig, ax = plt.subplots(figsize=(15, 8.5))
for _, line in net.line.iterrows():
    x1, y1 = coords[line["from_bus"]]
    x2, y2 = coords[line["to_bus"]]
    is_cable = "Kablo" in line["name"]
    ax.plot([x1, x2], [y1, y2], linestyle="--" if is_cable else "-",
            color="#1f77b4" if is_cable else "#444444", linewidth=2.2, zorder=1)
    ax.text((x1+x2)/2, (y1+y2)/2 + 0.25, f"{line['length_km']} km",
            fontsize=7.5, color="#1f77b4" if is_cable else "#444444", ha="center")

for _, trafo in net.trafo.iterrows():
    x1, y1 = coords[trafo["hv_bus"]]
    x2, y2 = coords[trafo["lv_bus"]]
    ax.plot([x1, x2], [y1, y2], linestyle="-", color="darkorange", linewidth=2.5, zorder=2)
    mx, my = (x1+x2)/2, (y1+y2)/2
    ax.scatter([mx], [my], s=180, facecolors="white", edgecolors="darkorange", linewidth=2.2, zorder=3)
    ax.text(mx, my - 0.55, f"TR ({trafo['sn_mva']} MVA)", fontsize=7.5, color="darkorange", fontweight="bold", ha="center")

for idx, (x, y) in coords.items():
    vn = net.bus.loc[idx, "vn_kv"]
    c = "crimson" if vn == 154.0 else ("navy" if vn == 33.0 else "purple")
    ax.scatter([x], [y], s=140, color=c, zorder=4)
    ax.text(x, y + 0.45, f"B{idx} ({vn} kV)", fontsize=8.5, fontweight="bold", ha="center", color=c)

for _, load in net.load.iterrows():
    x, y = coords[load["bus"]]
    ax.annotate(f"Yük: {load['p_mw']} MW", xy=(x, y), xytext=(x, y - 0.9),
                arrowprops=dict(arrowstyle="->", color="brown", lw=1.2),
                fontsize=7.5, color="brown", ha="center")

for _, sgen in net.sgen.iterrows():
    x, y = coords[sgen["bus"]]
    lbl = f"{'Rüzgar' if sgen['type']=='WP' else 'PV'}: {sgen['p_mw']} MW"
    ax.annotate(lbl, xy=(x, y), xytext=(x + 0.85, y + 1.15),
                arrowprops=dict(arrowstyle="->", color="forestgreen", lw=1.5),
                fontsize=8, fontweight="bold", color="forestgreen", ha="center",
                bbox=dict(boxstyle="round,pad=0.2", fc="#e8f5e9", ec="forestgreen", lw=1))

legend_items = [
    mlines.Line2D([], [], color="crimson", marker="o", linestyle="None", markersize=8, label="154 kV İletim Barası (Slack)"),
    mlines.Line2D([], [], color="navy", marker="o", linestyle="None", markersize=8, label="33 kV OG Barası"),
    mlines.Line2D([], [], color="purple", marker="o", linestyle="None", markersize=8, label="0.4 kV AG Barası"),
    mlines.Line2D([], [], color="#444444", linestyle="-", linewidth=2, label="33 kV Havai Hat"),
    mlines.Line2D([], [], color="#1f77b4", linestyle="--", linewidth=2, label="33 kV Yeraltı Kablosu"),
    mlines.Line2D([], [], color="darkorange", marker="o", markerfacecolor="white", markersize=9, label="Güç / Dağıtım Trafosu"),
]
ax.legend(handles=legend_items, loc="upper left", fontsize=8.5, frameon=True)
ax.set_xlim(-0.5, 14.0)
ax.set_ylim(2.2, 18.2)
ax.axis("off")
ax.set_title("14 Baralı OG (33 kV) + AG (0.4 kV) PV ve Rüzgar Entegreli Dağıtım Şebekesi Tek Hat Şeması", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("tek_hat_semasi.png", dpi=300)
plt.show()

# ==============================================================================
# 3. ANALİZ 1: PV + RÜZGAR DEVRE DIŞI vs DEVREDE (%100 Yük)
# ==============================================================================
net.sgen["in_service"] = False
net.load["scaling"] = 1.0
pp.runpp(net)
res_bus_off = net.res_bus.copy()
res_line_off = net.res_line.copy()
kayip_off = net.res_line["pl_mw"].sum() + net.res_trafo["pl_mw"].sum()

net.sgen["in_service"] = True
pp.runpp(net)
res_bus_on = net.res_bus.copy()
res_line_on = net.res_line.copy()
kayip_on = net.res_line["pl_mw"].sum() + net.res_trafo["pl_mw"].sum()

# B0 Slack barasını (1.02 pu sabit referans) aşırı gerilim olarak göstermemek için ayrıştırma
def gerilim_durumu(vm_series, slack_idx):
    dusum = np.maximum(0.0, (1.0 - vm_series) * 100).round(2).astype(str)
    yukselme = np.maximum(0.0, (vm_series - 1.0) * 100).round(2).astype(str)
    dusum.loc[slack_idx] = "Slack Ref."
    yukselme.loc[slack_idx] = "Slack Ref."
    return dusum, yukselme

dusum_off, yukselme_off = gerilim_durumu(res_bus_off["vm_pu"], b0)
dusum_on, yukselme_on   = gerilim_durumu(res_bus_on["vm_pu"], b0)

karsilastirma_df = pd.DataFrame({
    "Bara Adı": net.bus["name"],
    "kV": net.bus["vn_kv"],
    "V_pu (YOK)": res_bus_off["vm_pu"].round(4),
    "Ger. Düşümü % (YOK)": dusum_off,
    "Ger. Yükselmesi % (YOK)": yukselme_off,
    "V_pu (VAR)": res_bus_on["vm_pu"].round(4),
    "Ger. Düşümü % (VAR)": dusum_on,
    "Ger. Yükselmesi % (VAR)": yukselme_on,
})

print("=== 1. BARA GERİLİMLERİ, GERİLİM DÜŞÜMÜ VE GERİLİM YÜKSELMESİ (%100 YÜK) ===")
print("Not: B0 (154 kV) Slack barası 1.02 p.u. sabit referanstır; B1-B13 için 1.00 p.u. baz alınmıştır.")
print(karsilastirma_df.to_string(index=True))
print(f"\nToplam Aktif Güç Kaybı (PV+Rüzgar YOK): {kayip_off:.4f} MW")
print(f"Toplam Aktif Güç Kaybı (PV+Rüzgar VAR): {kayip_on:.4f} MW (Kayıp Azalması: %{((kayip_off-kayip_on)/kayip_off)*100:.1f})\n")

hat_df = pd.DataFrame({
    "Hat Adı": net.line["name"],
    "km": net.line["length_km"],
    "I_YOK (kA)": res_line_off["i_ka"].round(4),
    "I_VAR (kA)": res_line_on["i_ka"].round(4),
    "Yük. YOK (%)": res_line_off["loading_percent"].round(2),
    "Yük. VAR (%)": res_line_on["loading_percent"].round(2),
    "P_from YOK (MW)": res_line_off["p_from_mw"].round(3),
    "P_from VAR (MW)": res_line_on["p_from_mw"].round(3),
    "Q_from VAR (MVAr)": res_line_on["q_from_mvar"].round(3)
})
print("=== 2. HAT AKIMLARI VE ÇİFT YÖNLÜ GÜÇ AKIŞI DAĞILIMI (YOK vs VAR) ===")
print(hat_df.to_string(index=False))

# ==============================================================================
# 4. ANALİZ 2: FARKLI YÜKLENME KOŞULLARI (%40, %60, %80, %90)
# ==============================================================================
yuklenme_oranlari = [0.40, 0.60, 0.80, 0.90]
yuk_sonuclari = []
og_ag_mask = net.bus.index != b0  # B0 Slack harici tüm OG ve AG baraları (B1-B13)

for oran in yuklenme_oranlari:
    net.load["scaling"] = oran

    net.sgen["in_service"] = False
    pp.runpp(net)
    og_ag_off = net.res_bus.loc[og_ag_mask, "vm_pu"]
    min_v_off = og_ag_off.min()
    max_v_off = og_ag_off.max()
    max_drop_off = max(0.0, (1.0 - min_v_off) * 100)
    loss_off = net.res_line["pl_mw"].sum() + net.res_trafo["pl_mw"].sum()

    net.sgen["in_service"] = True
    pp.runpp(net)
    og_ag_on = net.res_bus.loc[og_ag_mask, "vm_pu"]
    min_v_on = og_ag_on.min()
    max_v_on = og_ag_on.max()
    max_drop_on = max(0.0, (1.0 - min_v_on) * 100)
    max_rise_on = max(0.0, (max_v_on - 1.0) * 100)
    loss_on = net.res_line["pl_mw"].sum() + net.res_trafo["pl_mw"].sum()

    yuk_sonuclari.append({
        "Yüklenme": f"%{int(oran*100)}",
        "Min V YOK (pu)": round(min_v_off, 4),
        "Max V YOK (pu)": round(max_v_off, 4),
        "Max Düşüm YOK (%)": round(max_drop_off, 2),
        "Min V VAR (pu)": round(min_v_on, 4),
        "Max V VAR (pu)": round(max_v_on, 4),
        "Max Düşüm VAR (%)": round(max_drop_on, 2),
        "Max Yükselme VAR (%)": round(max_rise_on, 2),
        "Kayıp YOK (MW)": round(loss_off, 4),
        "Kayıp VAR (MW)": round(loss_on, 4)
    })

yuk_df = pd.DataFrame(yuk_sonuclari)
print("\n=== 3. FARKLI YÜKLENME KOŞULLARI (%40-%90) GERİLİM DÜŞÜMÜ VE YÜKSELMESİ (B1–B13) ===")
print(yuk_df.to_string(index=False))

# ==============================================================================
# 5. ANALİZ 3: 3 FAZLI KISA DEVRE ANALİZİ (IEC 60909)
# ==============================================================================
net.load["scaling"] = 1.0
net.sgen["in_service"] = True

sc.calc_sc(net, case="max", fault="3ph", ip=True, ith=True)
sc_max_df = net.res_bus_sc[["ikss_ka", "skss_mw", "ip_ka", "ith_ka"]].copy()
sc_max_df.columns = ["Ikss_max (kA)", "Skss_max (MVA)", "Ip_darbe (kA)", "Ith_termik (kA)"]

sc.calc_sc(net, case="min", fault="3ph")
sc_max_df["Ikss_min (kA)"] = net.res_bus_sc["ikss_ka"]
sc_max_df.insert(0, "Bara Adı", net.bus["name"])
sc_max_df.insert(1, "kV", net.bus["vn_kv"])

print("\n=== 4. 3 FAZLI KISA DEVRE ANALİZİ SONUÇLARI ===")
print(sc_max_df.round(3).to_string(index=False))

# ==============================================================================
# 6. KARŞILAŞTIRMA GRAFİKLERİ (B0 Slack Hariç OG & AG Baralarına Odaklı)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

og_ag_idx = net.bus.index[og_ag_mask]
axes[0].plot(og_ag_idx, res_bus_off.loc[og_ag_mask, "vm_pu"], marker="o", color="crimson", linewidth=2, label="PV + Rüzgar Devre Dışı")
axes[0].plot(og_ag_idx, res_bus_on.loc[og_ag_mask, "vm_pu"], marker="s", color="forestgreen", linewidth=2, label="PV + Rüzgar Devrede")
axes[0].axhline(1.00, color="navy", linestyle=":", linewidth=1, label="Nominal Gerilim (1.00 pu)")
axes[0].axhline(0.95, color="gray", linestyle="--", linewidth=1, label="Alt Sınır (0.95 pu)")
axes[0].set_xticks(og_ag_idx)
axes[0].set_xticklabels([f"B{i}" for i in og_ag_idx])
axes[0].set_ylabel("Bara Gerilimi (p.u.)")
axes[0].set_title("OG ve AG Baraları (B1–B13) Gerilim Profili (%100 Yük)\n(B0 Slack Barası 1.02 pu Sabit Referanstır)")
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=8.5)

x2 = np.arange(len(yuklenme_oranlari))
width = 0.28
axes[1].bar(x2 - width, yuk_df["Max Düşüm YOK (%)"], width, color="crimson", label="Max Gerilim Düşümü (YOK)")
axes[1].bar(x2, yuk_df["Max Düşüm VAR (%)"], width, color="forestgreen", label="Max Gerilim Düşümü (VAR)")
axes[1].bar(x2 + width, yuk_df["Max Yükselme VAR (%)"], width, color="royalblue", label="Max Gerilim Yükselmesi (VAR)")
axes[1].set_xticks(x2)
axes[1].set_xticklabels(yuk_df["Yüklenme"])
axes[1].set_ylabel("Gerilim Değişimi (%)")
axes[1].set_title("Yüklenme Oranlarına Göre Gerilim Düşümü & Yükselmesi")
axes[1].grid(True, axis="y", alpha=0.3)
axes[1].legend(fontsize=8.5)

plt.tight_layout()
plt.savefig("analiz_grafikleri_v2.png", dpi=300)
plt.show()
```
